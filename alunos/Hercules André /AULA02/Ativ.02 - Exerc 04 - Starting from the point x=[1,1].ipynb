Starting from the point x=[1.0,1.0], implement the gradient descent and Newton method to find the minimum of this function.

Vamos assumir a função:
f(x) = (x1 - 3)^2 + (x2 - 2)^2

O ponto inicial:
x = [1.0,1.0]

Importar a biblioteca

```julia
using Pkg
Pkg.add("LinearAlgebra")
```

```julia
using LinearAlgebra
```

```julia
using Plots
```

```julia
#Função
function f(x)
    return (x[1] - 3)^2 + (x[2] - 2)^2
end

# Gradiente
function grad_f(x)
    return [2*(x[1] - 3), 2*(x[2] - 2)]
end

# Hessiana
function hess_f(x)
    return [2.0 0.0;
            0.0 2.0]
end


```

Gradient Descent

```julia
function gradient_descent(x0; α=0.1, tol=1e-6, max_iter=100)
    x = x0
    path = [x]

    for k in 1:max_iter
        g = grad_f(x)
        
        if norm(g) < tol
            break
        end
        
        x = x - α * g
        push!(path, x)
    end

    return x, path
end
```

Executar

```julia
x0 = [1.0, 1.0]

x_gd, path_gd = gradient_descent(x0)

println("Resultado Gradient Descent: ", x_gd)
```

Método de Newton

```julia
function newton_method(x0; tol=1e-6, max_iter=20)
    x = x0
    path = [x]

    for k in 1:max_iter
        g = grad_f(x)
        H = hess_f(x)

        if norm(g) < tol
            break
        end

        x = x - H \ g
        push!(path, x)
    end

    return x, path
end
```

Executar

```julia
x_newton, path_newton = newton_method(x0)

println("Resultado Newton: ", x_newton)
```
