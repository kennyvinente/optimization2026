Gradient Descent with Backtracking Line Search: 
Given function: f(x1,x2,x3) =  x3 * log(exp(x1/x3) + exp(x2/x3)) + (x3 - 2)^2 + xp(1 / (x1 + x2))
Define the function in the cell below:

IMPORTAR BIBLIOTECA

```julia
using LinearAlgebra
```

FUNÇÃO

```julia
function f(x)
    x1, x2, x3 = x
    
    if x1 + x2 <= 0 || x3 <= 0
        return Inf
    end
    term1 = x3 * log(exp(x1/x3) + exp(x2/x3))
    term2 = (x3 - 2)^2
    term3 = exp(1 / (x1 + x2))
    
    return term1 + term2 + term3
end
```

GRADIENTE

```julia
function grad_f(x)
    x1, x2, x3 = x
    
    if x1 + x2 <= 0 || x3 <= 0
        return [Inf, Inf, Inf]
    end
    
    exp1 = exp(x1/x3)
    exp2 = exp(x2/x3)
    denom = exp1 + exp2
    
# Derivadas
    df_dx1 = exp1 / denom + (-1/(x1+x2)^2)*exp(1/(x1+x2))
    df_dx2 = exp2 / denom + (-1/(x1+x2)^2)*exp(1/(x1+x2))
    
    df_dx3 = log(denom) 
    df_dx3 += x3 * ( (-x1/x3^2)*exp1 + (-x2/x3^2)*exp2 ) / denom
    df_dx3 += 2*(x3 - 2)
    
    return [df_dx1, df_dx2, df_dx3]
end
    
```

BACKTRACKING

```julia
function backtracking(f, grad_f, x, d; α=0.25, β=0.5)
    t = 1.0
    
    while f(x + t*d) > f(x) + α*t*dot(grad_f(x), d)
        t *= β
    end
    
    return t
end
```

GRADIENT DESCENT

```julia
function gradient_descent(f, grad_f, x0; tol=1e-6, max_iter=1000)
    
    x = x0
    history = [x0]
    
    for k in 1:max_iter
        
        g = grad_f(x)
        
        if norm(g) < tol
            println("Convergiu em $k iterações")
            break
        end
        
        d = -g
        
        t = backtracking(f, grad_f, x, d)
        
        x = x + t*d
        
        push!(history, x)
    end
    
    return x, history
end
```

RESOLVER

```julia
x0 = [1.0, 1.0, 1.0]

x_opt, hist = gradient_descent(f, grad_f, x0)

println("Solução encontrada:")
println(x_opt)

println("Valor da função:")
println(f(x_opt))
```
