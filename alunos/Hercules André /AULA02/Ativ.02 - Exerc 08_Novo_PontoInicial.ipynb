Ponto inicial: x = [-2.0,2.0]

IMPORTAR BIBLIOTECA

```julia
using LinearAlgebra
```

FUNÇÃO

```julia
f(x) = x[1]^2 + x[2]^2
```

GRADIENTE

```julia
function grad_f(x)
    return [2*x[1], 2*x[2]]
end
```

PONTO INICIAL

```julia
x0 = [-2.0, 2.0]
```

MÉTODO GRADIENTE DESCENDENTE


```julia
function gradient_descent(x0; α=0.01, tol=1e-6, max_iter=1000)
    x = copy(x0)
    
    for k in 1:max_iter
        g = grad_f(x)
        
        println("Iter $k: x = $x, f(x) = $(f(x))")
        
        if norm(g) < tol
            println("Convergiu!")
            break
        end
        
        x = x - α * g
    end
    
    return x
end
```
