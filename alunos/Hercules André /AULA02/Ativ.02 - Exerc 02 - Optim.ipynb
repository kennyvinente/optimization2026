Atividade 02 - Exercicio 02 (Solving the FONC)

Given the function z = (x1 -1)^4 + (x2 -2)^4

using the optim.jl package, find the minimum of z.


Importar a biblioteca


```julia
using Pkg
Pkg.add("Optim")
```

```julia
using Optim
```

Definir a Função

```julia
function z(x)
    return (x[1] - 1)^4 + (x[2] - 2)^4
end
```

Ponto Inicial

```julia
x0 = [0.0, 0.0]
```

Resolver

```julia
result = optimize(z, x0)
```

Resultado

```julia
println("Solução ótima: ", Optim.minimizer(result))
println("Valor mínimo: ", Optim.minimum(result))
```
