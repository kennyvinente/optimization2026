IMPORTAR A BIBLIOTECA

```julia
using LinearAlgebra
```

FUNÇÃO

```julia
function f(x)
    x1, x2, x3 = x
    
    return log(exp(x1) + exp(x2)) + 1/(x1 + x2) + (x3 - 2)^2
end
```

IMPLEMENTAR O GRADIENTE

```julia
function grad_f(x)
    x1, x2, x3 = x
    
    # Termo comum
    denom = exp(x1) + exp(x2)
    
    # Derivadas
    df_dx1 = exp(x1)/denom - 1/(x1 + x2)^2
    df_dx2 = exp(x2)/denom - 1/(x1 + x2)^2
    
    df_dx3 = log(denom) - (exp(x1) + exp(x2)) / (x3 * denom) + 2*(x3 - 2)
    
    return [df_dx1, df_dx2, df_dx3]
end
```

TESTAR O GRADIENTE

```julia
x = [1.0, 1.0, 1.0]

grad_f(x)
```
