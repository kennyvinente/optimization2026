Gradient Descent com Backtracking Método de Newton Tolerância: 1e-5 para a norma do gradiente

--------------------------------------------------------

IMPORTAR A BIBLIOTECA

```julia
using LinearAlgebra
```

FUNÇÃO

```julia
function f(x)
    x1, x2, x3 = x
    return x3 * log(exp(x1/x3) + exp(x2/x3)) + (x3 - 2)^2 + exp(-(x1 + x2))
end
```

GRADIENTE

```julia
function grad_f(x)
    x1, x2, x3 = x

    A = exp(x1/x3)
    B = exp(x2/x3)
    S = A + B

    df_dx1 = A/S - exp(-(x1 + x2))
    df_dx2 = B/S - exp(-(x1 + x2))
    df_dx3 = log(S) - (x1*A + x2*B)/(x3*S) + 2*(x3 - 2)

    return [df_dx1, df_dx2, df_dx3]
end
```

BACKTRACKING LINE SEARCH

```julia
function backtracking(f, grad_f, x, d; t=1.0, α=0.4, β=0.5)
    while f(x + t*d) > f(x) + α*t*dot(grad_f(x), d)
        t *= β
    end
    return t
end
```

MÉTODO BFGS

```julia
function BFGS(x0; tol=1e-5, max_iter=1000)
    x = x0
    n = length(x)
    H = Matrix{Float64}(I, n, n)  # Aproximação da inversa da Hessiana

    for k in 1:max_iter
        g = grad_f(x)

        if norm(g) < tol
            println("Convergiu em $k iterações")
            return x
        end

        # Direção de descida
        d = -H * g

        # Line search
        t = backtracking(f, grad_f, x, d)

        # Atualização do ponto
        x_new = x + t*d

        # Vetores s e y
        s = x_new - x
        y = grad_f(x_new) - g

        # Condição de curvatura
        if dot(y, s) > 1e-10
            ρ = 1.0 / dot(y, s)
            I_n = Matrix{Float64}(I, n, n)

            H = (I_n - ρ * s * y') * H * (I_n - ρ * y * s') + ρ * (s * s')
        end

        x = x_new
    end

    println("Máximo de iterações atingido")
    return x
end
```

PONTO INICIAL

```julia
x0 = [1.0, 1.0, 1.0]
```

EXECUÇÃO

```julia
x_bfgs = BFGS(x0)

println("Solução BFGS: ", x_bfgs)
println("Valor da função: ", f(x_bfgs))
```
