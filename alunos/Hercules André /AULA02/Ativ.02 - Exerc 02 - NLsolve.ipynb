Atividade 02 - Exercicio 02 (Solving the FONC)

Given the function z = (x1 -1)^4 + (x2 -2)^4

using the NLsolve.jl package, solve the FONC problem for this function

Importar a Biblioteca

```julia
using Pkg
Pkg.add("NLsolve")
```

```julia
using NLsolve
```

Definir sistema FONC

```julia
function fonc!(F, x)
    F[1] = 4*(x[1] - 1)^3
    F[2] = 4*(x[2] - 2)^3
end
```

Chute Inicial

```julia
x0 = [0.0, 0.0]
```

Resolver o sistema

```julia
sol = nlsolve(fonc!,x0)
```

Resultado

```julia
println("Solução FONC:", sol.zero)
```
