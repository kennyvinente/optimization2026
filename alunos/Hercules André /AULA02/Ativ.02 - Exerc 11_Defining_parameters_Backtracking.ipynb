Defining parameters for backtracking search
alpha = 0.4
beta = 0.5
eps = 1e-5

IMPORTAR BIBLIOTECAS

```julia
using LinearAlgebra
```

FUNÇÃO

```julia
function f(x)
    x1, x2, x3 = x
    return log(exp(x1) + exp(x2)) + 1/(x1 + x2) + (x3 - 2)^2
end
```

GRADIENTE

```julia
function grad_f(x)
    x1, x2, x3 = x
    
    denom = exp(x1) + exp(x2)
    
    df_dx1 = exp(x1)/denom - 1/(x1 + x2)^2
    df_dx2 = exp(x2)/denom - 1/(x1 + x2)^2
    df_dx3 = 2*(x3 - 2)
    
    return [df_dx1, df_dx2, df_dx3]
end
```

DEFINIR O BACKTRACKING

```julia
function backtracking(f, grad_f, x, d; alpha=0.4, beta=0.5)
    
    t = 1.0
    
    while f(x + t*d) > f(x) + alpha * t * dot(grad_f(x), d)
        t = beta * t
    end
    
    return t
end
```

GRADIENT DESCENT COM BACKTRACKING

```julia
function gradient_descent_bt(f, grad_f, x0; alpha=0.4, beta=0.5, eps=1e-5, max_iter=1000)
    
    x = x0
    history = [x0]
    
    for k in 1:max_iter
        
        g = grad_f(x)
        
        # critério de parada
        if norm(g) < eps
            println("Convergiu em ", k, " iterações")
            break
        end
        
        d = -g
        
        t = backtracking(f, grad_f, x, d, alpha=alpha, beta=beta)
        
        x = x + t*d
        
        push!(history, x)
    end
    
    return x, history
end
```

EXECUTAR

```julia
x0 = [1.0, 1.0, 1.0]

x_min, history = gradient_descent_bt(f, grad_f, x0)

println("Mínimo encontrado: ", x_min)
```
