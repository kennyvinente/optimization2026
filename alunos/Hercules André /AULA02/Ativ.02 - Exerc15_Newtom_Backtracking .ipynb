Newton com backtracking line search e critério λ²/2 ≤ ε

IMPORTAR A BIBLIOTECA

```julia
using LinearAlgebra
```

Para demonstração iremos utilizar a primeira função da Atividade.

f(x)=(x1​−3)2+(x2​−2)2

FUNÇÃO

```julia
f(x) = (x[1] - 3)^2 + (x[2] - 2)^2
```

GRADIENTE

```julia
grad_f(x) = [2*(x[1] - 3), 2*(x[2] - 2)]
```

HESSIANA

```julia
hess_f(x) = [2 0; 0 2]
```

BACKTRAKING

```julia
function backtracking(x, Δx; α=0.25, β=0.5)
    t = 1.0
    while f(x + t*Δx) > f(x) + α*t*dot(grad_f(x), Δx)
        t *= β
    end
    return t
end
```

MÉTODO DE NEWTON

```julia
function newton(x0; ε=1e-6, max_iter=100)
    x = x0

    for k in 1:max_iter
        g = grad_f(x)
        H = hess_f(x)

        Δx = - H \ g
        λ2 = dot(g, H \ g)

        println("Iter $k | x = $x | f(x) = $(f(x))")

        if λ2/2 <= ε
            println("Convergiu em $k iterações")
            return x
        end

        t = backtracking(x, Δx)
        x = x + t * Δx
    end

    return x
end
```

RESULTADO

```julia
x0 = [1.0, 1.0]
xmin = newton(x0)

println("Mínimo: ", xmin)
```
