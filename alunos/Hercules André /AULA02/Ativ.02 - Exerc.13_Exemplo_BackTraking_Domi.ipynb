IMPORTAR A BIBLIOTECA

```julia
using LinearAlgebra
```

FUNÇÃO

```julia
function f(x)
    x1, x2, x3 = x
    return log(exp(x1) + exp(x2)) + 1/(x1 + x2) + (x3 - 2)^2
end
```

DEFININDO O GRADIENTE

```julia
function grad_f(x)
    x1, x2, x3 = x
    
    denom = exp(x1) + exp(x2)
    
    df_dx1 = exp(x1)/denom - 1/(x1 + x2)^2
    df_dx2 = exp(x2)/denom - 1/(x1 + x2)^2
    df_dx3 = 2*(x3 - 2)
    
    return [df_dx1, df_dx2, df_dx3]
end
```

FUNÇÃO DOMÍNIO

```julia
function in_domain(x)
    x1, x2, x3 = x
    
    # condição do domínio
    if abs(x1 + x2) < 1e-8
        return false
    end
    
    return true
end
```

PARAMETROS

```julia
alpha = 0.4
beta = 0.5
eps = 1e-5
```

BACKTRACKING COM DOMÍNIO

```julia
function backtracking_domain(f, grad_f, x, d; alpha=0.4, beta=0.5)
    
    t = 1.0
    
    while true
        
        x_new = x + t*d
        
        # 🔴 Garantir domínio
        if !in_domain(x_new)
            t = beta * t
            continue
        end
        
        # 🔵 Condição de Armijo
        if f(x_new) > f(x) + alpha * t * dot(grad_f(x), d)
            t = beta * t
            continue
        end
        
        break
    end
    
    return t
end
```

GRADIENT DESCENT

```julia
function gradient_descent_bt_domain(f, grad_f, x0; alpha=0.4, beta=0.5, eps=1e-5, max_iter=1000)
    
    x = x0
    history = [x0]
    
    for k in 1:max_iter
        
        g = grad_f(x)
        
        println("Iteração ", k)
        println("x = ", x)
        println("||grad|| = ", norm(g))
        println("---------------------------")
        
        # critério de parada
        if norm(g) < eps
            println("Convergiu!")
            break
        end
        
        d = -g
        
        t = backtracking_domain(f, grad_f, x, d, alpha=alpha, beta=beta)
        
        x = x + t*d
        
        push!(history, x)
    end
    
    return x, history
end
```

SIMULAR

```julia
x0 = [3.0, 4.0, 5.0]

x_min, history = gradient_descent_bt_domain(f, grad_f, x0)

println("Mínimo encontrado: ", x_min)
```
