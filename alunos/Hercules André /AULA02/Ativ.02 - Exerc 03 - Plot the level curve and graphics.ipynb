Atividade 02 - Exercicio 03 
Plot the level curve and the graphics of f(x) = (x1 - 3)^2 + (x2 - 2)^2

Importando a Biblioteca

```julia
using Plots
```

Definição da Função

```julia
f(x1, x2) = (x1 - 3)^2 + (x2 - 2)^2
```

Criando o Grid

```julia
x1 = range(-2, 8, length=100)
x2 = range(-2, 8, length=100)
X1 = repeat(x1', length(x2),1)
X2 = repeat(x2, 1, length(x1))

Z = f.(X1, X2)
```

Plots das curvas de Nível

```julia
contour(x1, x2, Z,
    title = "Curvas de Nível",
    xlabel = "x1",
    ylabel = "x2")
```

Grafico 3 D

```julia
surface(x1, x2, Z,
    title="f(x1,x2) = (x1-3)^2 + (x2-2)^2",
    xlabel="x1",
    ylabel="x2",
    zlabel="f(x)",
)
```
