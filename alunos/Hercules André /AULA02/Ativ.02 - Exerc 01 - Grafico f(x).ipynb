FONC Example:

Plot the graphics de f(x) = (x1 - 3)^2 + (x2 + 4)^2

Gradiente (FONC): 
     ∇ f(x) = [2(x1 - 3)
     
              2(x2 - 4)]
  


```julia
Importando a biblioteca
```

```julia
using Pkg
Pkg.add("Plots")
```

```julia
using Plots
```

Definir a Função

```julia
f(x1, x2) = (x1 - 3 )^2 + (x2 - 4)^2
```

Grid

```julia
x1 = range(-5, 10, length=100)
x2 = range(-10 , 5,length=100)
```

```julia
Z = [f(x, y) for x in x1, y in x2]
```

Plotar 3 D

```julia
surface(x1, x2, Z,
    title="f(x1,x2) = (x1-3)^2 + (x2-4)^2",
    xlabel="x1",
    ylabel="x2",
    zlabel="f(x)",
)
```

Plot de Curvas de nível(contorno)

```julia
contour(x1, x2, Z,
    title="Curvas de nível",
    xlabel="x1",
    ylabel="x2",
)
```

Marcar o ponto ótimo

```julia
scatter!([3], [-4], color=:red, label="Mínimo")
```
