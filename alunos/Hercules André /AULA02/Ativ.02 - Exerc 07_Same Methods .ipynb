```julia
function gradient_descent(x0; α=0.01, tol=1e-6, max_iter=1000)
    x = copy(x0)
    
    for k in 1:max_iter
        g = grad_f(x)
        
        println("Iter $k: x = $x, f(x) = $(f(x))")
        
        if norm(g) < tol
            println("Convergiu!")
            break
        end
        
        x = x - α * g
    end
    
    return x
end
```
