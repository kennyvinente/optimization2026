Gradient Descent com Backtracking
Método de Newton
Tolerância: 1e-5 para a norma do gradiente

----------------------------------------------------

IMPORTAR BIBLIOTECA

```julia
using LinearAlgebra
```

FUNÇÃO

```julia
function f(x)
    x1, x2, x3 = x
    return x3 * log(exp(x1/x3) + exp(x2/x3)) + (x3 - 2)^2 + 1/(exp(x1 + x2))
end
```

GRADIENTE

```julia
function grad_f(x)
    x1, x2, x3 = x

    A = exp(x1/x3)
    B = exp(x2/x3)
    S = A + B

    df_dx1 = (x3 * (A/(x3*S))) - exp(-(x1+x2))
    df_dx2 = (x3 * (B/(x3*S))) - exp(-(x1+x2))

    df_dx3 = log(S) - (x1*A + x2*B)/(x3*S) + 2*(x3 - 2)

    return [df_dx1, df_dx2, df_dx3]
end
```

HESSIANA

```julia
function hess_f(x)
    x1,x2,x3 = x

    A = exp(x1/x3)
    B = exp(x2/x3)
    S = A + B

    # aproximação simplificada
    H = zeros(3,3)

    #termos principais
    H[1,1] = A/(x3*S) - exp(-(x1+x2))
    H[2,2] = A/(x3*S) - exp(-(x1+x2))
    H[1,2] = -exp(-(x1+x2))
    H[2,1] = H[1,2]

    H[3,3] = 2.0

    return H
end
```

BACKTRACKING LINE SEARCH

```julia
function backtracking(f, grad_f, x, d; t=1.0, α=0.4, β=0.5)
    while f(x + t*d) > f(x) + α*t*dot(grad_f(x), d)
        t *= β
    end
    return t
end
```

i -  GRADIENT DESCENT

```julia
function gradient_descent(x0; tol=1e-5, max_iter=1000)
    x = x0

    for k in 1:max_iter
        g = grad_f(x)

        if norm(g) < tol
            println("Convergiu em $k iterações")
            return x
        end

        d = -g
        t = backtracking(f, grad_f, x, d)

        x = x + t*d
    end

    return x
end
```

PONTO INICIAL

```julia
x0 = [1.0, 1.0, 1.0]
```

i - EXECUTAR O GRADIENT DESCENT

```julia
x_gd = gradient_descent(x0)
println("Solução GD: ", x_gd)
println("f(x): ", f(x_gd))
```

ii - MÉTODO DE NEWTON

```julia
function newton_method(x0; tol=1e-5, max_iter=100)
    x = x0

    for k in 1:max_iter
        g = grad_f(x)

        if norm(g) < tol
            println("Convergiu em $k iterações")
            return x
        end

        H = hess_f(x)
        d = -H \ g

        t = backtracking(f, grad_f, x, d)

        x = x + t*d
    end

    return x
end
```

ii - EXECUTAR NEWTON

```julia
x_newton = newton_method(x0)
println("Solução Newton: ", x_newton)
println("f(x): ", f(x_newton))
```
