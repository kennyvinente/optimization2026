IMPORTAR A BIBLIOTECA

```julia
using LinearAlgebra
```

FUNÇÃO

```julia
function f(x)
    return (x[1] - 3)^2 + (x[2] - 2)^2
end
```

GRADIENTE

```julia
function grad_f(x)
    return [2*(x[1] - 3), 2*(x[2] - 2)]
end
```

HESSIANA

```julia
function hess_f(x)
    return [2.0 0.0;
            0.0 2.0]
end
```

BACKTRACKING LINE SEARCH

```julia
function backtracking(x, Δx; α=0.1, β=0.5)
    t = 1.0
    
    # Garantir domínio (aqui assumimos que sempre está no domínio)
    # Se tiver restrição, você coloca a condição aqui
    
    # Condição de Armijo
    while f(x + t*Δx) > f(x) + α * t * dot(grad_f(x), Δx)
        t = β * t
    end
    
    return t
end
```

```julia
MÉTODO DE NEWTON
```

```julia
function newton_backtracking(x0; tol=1e-6, max_iter=100)
    x = x0
    
    for k in 1:max_iter
        
        g = grad_f(x)
        H = hess_f(x)
        
        # Direção de Newton
        Δx = - H \ g
        
        # Critério de parada
        if norm(g) < tol
            println("Convergiu em $k iterações")
            return x
        end
        
        # Backtracking
        t = backtracking(x, Δx)
        
        # Atualização
        x = x + t * Δx
        
        println("Iter $k | x = $x | f(x) = $(f(x))")
    end
    
    return x
end
```

EXECUTAR

```julia
x0 = [1.0, 1.0]

resultado = newton_backtracking(x0)

println("Mínimo encontrado: ", resultado)
```
