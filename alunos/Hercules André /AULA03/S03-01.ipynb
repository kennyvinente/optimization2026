```julia
#Importar a Biblioteca

using LinearAlgebra
using Printf

#Função F(x)

function F(x)

    f1 = (x[1] + 1)^2 + x[2]^2 - 2
    f2 = exp(x[1]) + x[2]^3 - 2

    return [f1; f2]

end

#Matriz Jacobiana

function J(x)

    return [
        2*(x[1] + 1)    2*x[2];
        exp(x[1])       3*x[2]^2
        ]
end
```

```julia

# Condição Inicial
x = [1.0; 1.0]

#Tolerância
tol = 1e-15

#Número de Interações
max_iter = 100
```

```julia
# Método de Newton
println("Método de Newton Multivariado\n")

for k in 1:max_iter

    #Calcula F(x)
    Fx = F(x)

    #Calcula Jacobiana
    Jx = J(x)

    #Resolve o sistema linear
    #J(x)s = -F(x)
    s = -Jx \ Fx

    #Atualiza a solução
    x_new = x + s

    #Norma do passo
    erro = norm(s)

    @printf("Iteração %2d\n", k)
    @printf("x1 = %.15f\n", x_new[1])
    @printf("x2 = %.15f\n", x_new[2])
    @printf("Erro = %.15e\n\n", erro)

    #Critério de parada
    if erro < tol
        println("Convergência alcançada!")
        x = x_new
        break
    end

    x = x_new

end
```

```julia
#Resultado

println("solução aproximada:\n")

@printf("x1 = %.15f\n", x[1])
@printf("x2 = %.15f\n", x[2])
```
