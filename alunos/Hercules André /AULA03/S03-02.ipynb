```julia
# importar biblioteca
using Printf

#Função do problema 
function F(x)
    return x^2 - 2
end
```

```julia
#Parâmetros
# Aproximação inicial 
x0 =  2.0

# Parametro tau
tau = 1e-7

#Tolerância
tol = 1e-15

#número de interações 
max_iter = 100
```

```julia
#Método Quasi-Newton
function finite_difference_newton(F, x0, tau, tol, max_iter)

    xk = x0
    k = 0

    println("Interação |       xk          |     F(xk)")
    println("---------------------------------------------------")

    while abs(F(xk)) > tol && k < max_iter
        
        #Parametro s
        if abs(xk) > 1
            s = tau * xk
        else
            s = tau
        end

        #fórmula do método

        x_next = xk - (s * F(xk)) / (F(xk + s) - F(xk))

        @printf("%5d              | %.15f   |  %.15e\n", k, xk, F(xk))

        #Atualiza

        xk = x_next
        k += 1
    end

    println("------------------------------------------------------")
    println("Solução aproximada:")
    println(xk)

    return xk

end
    
```

```julia
raiz = finite_difference_newton(F, x0, tau, tol, max_iter)
```

```julia
#Biblioteca
using Printf

#Função
function F(x)
    return x^2 - 2
end
```

```julia
#Aproximação inicial
x0 = 2.0

#Aproximação inicial da derivada
a0 = 1.0

#Tolerância
tol =1e-6

#Número de interações
max_iter = 100
```

```julia
#inicialização
xk = x0
ak = a0

println("Iteração\t xk\t\t F(xk)\t\t ak")

for k in 0:max_iter
    fx = F(xk)

    @printf("%d\t\t %.10f\t %.10f\t %.10f\n", k, xk, fx, ak)

    #Critério de parada
    if abs(fx) <= tol
        println("\nRaiz encontrada!")
        break
    end

    #Atualização de x
    x_next = xk - fx / ak

    #Atualização da derivada aproximada
    a_next = (F(xk) - F(x_next)) / (xk - x_next)

    #Atualizando as variãveis
    xk = x_next
    ak = a_next


end

println("\nRaiz aproximada = ", raiz)
```

```julia
#Bibliotecas

using LinearAlgebra
using Printf

# Função

function F(x)

    f1 = (x[1] + 1)^2 + x[2]^2 -2
    f2 = exp(x[1]) + x[2]^3 - 2

    return [f1,f2]

end
```

```julia
# Aproximação inicial 
x0 = [1.0 ; 1.0]

# Parametro tau
tau = 1e-7

#Tolerância
tol = 1e-10

#número de interações 
max_iter = 100
```

JACOBIANA APROXIMADA

```julia
function finite_difference_jacobian(F, x, tau)
    
    n = length(x)
    A = zeros(n,n)
    Fx = F(x)

    for j in 1:n

        #vetor canônico
        ej = zeros(n)
        ej[j] = 1.0

        #Construção de sj
         if abs(x[j]) >= 1
            sj = tau * x[j]
        elseif x[j] >= 0 && x[j] <= 1
            sj = tau
        else
            sj = -tau
        end

        #Aproximação da derivada
        A[:,j] = (F(x + sj*ej) - Fx)/sj

    end
    
    return A

end
```

```julia
#Método de Newton

function finite_difference_newton(F, x0, tau, tol, max_iter)

    x = copy(x0)

    println("Iteração        x1              x2              ||F(x)||")

    for k in 1:max_iter

        Fx = F(x)

        erro = norm(Fx)

         @printf("%3d        %12.8f    %12.8f    %.10e\n",
                k, x[1], x[2], erro)


        #Critério de parada

        if erro < tol

            println("\nConvergência atingida!")

            return x
        end

        #Jacobiana aproximada

        A = finite_difference_jacobian(F, x, tau)

        # Resolver o sistema linear

        d = -A \ Fx

        #Atualização

        x = x + d

    end

    println("\nNúmero  Máximo de iterações atingindo.")

    return x

end

        
```

```julia
#Resultado
x0 = [1.0; 1.0]

sol = finite_difference_newton(F, x0, tau, tol, max_iter)

println("\nSolução aproximada:")

println(sol)
```

```julia
#Bibliotecas

using LinearAlgebra
using Printf

#Função

function F(x)

    return [
        (x[1] + 1)^2 + x[2]^2 - 2
        exp(x[1]) + x[2]^3 - 2
    ]

end
```

```julia
#Parâmetros
#Vetor inicial

x =  [1.0;1.0]

#Tolerância

tol = 1e-7

#Número máximo de iterações
max_iter = 100

#Número de variáveis 
n = length(x)

# Aproximação inicial da Jacobiana
B = Matrix(I, n, n)
```

```julia
#Valores Iniciais
println("========== MÉTODO DE BROYDEN ==========")
println()

println("x0 =")
println(x)

println()
println("B0 = ")
println(B)
```

```julia
# Método da Secante Multivariada

for k = 1:max_iter
    println()
    println("==============================")
    println("Iteração: ", k)
    println("==============================")

    #Avaliar F(x)
    Fx = F(x)
    println()
    println("F(x) = ")
    println(Fx)

    #Verficar convergência
    if norm(Fx) < tol

        println()
        println("Convergência atingida!")

        break
    end

    #Resolver 
    #B*s = - F(x)

    s = -B \ Fx

    println()
    println("s = ")
    println(s)

    #Atualizar x

    x_new = x + s

    println()
    println("x novo = ")
    println(x_new)

    # Calcular y
    y = F(x_new) - Fx
    
    println()
    println("y = ")
    println(y)

    #Atualização de Broyden
    B = B + ((y - B*s) * transpose(s)) / (transpose(s)*s)

    println()
    println("Nova matriz B = ")
    println(B)

    # Atualizar x

    x = x_new

end
```

```julia
#Resultado

println()
println("========== RESULTADO FINAL ==========")

println()
println("Solução aproximada:")

println(x)

println()
println("F(x) final:")

println(F(x))
```
