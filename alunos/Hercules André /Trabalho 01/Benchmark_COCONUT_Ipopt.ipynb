```julia
using JuMP
using Ipopt
using XLSX
using Dates

# 1. Colagem dos dados copiados da sua tabela
dados_tabela = """
1001abelQLR2-AN301428094225.19460000abel.res
1002alkylQOR2-AN14712631-1.76500000alkyl.res
1003bearingLOR2-AN131212940
1004camcgeOOR2-AN2792421871331140(0.00000000)camcge.res
1005camcnsNOR2-AN2802431871341141(0.00000000)camcns.res
1006camshape100LQR2-AN199200100101696-4.28410000camshape100.res
1007camshape200LQR2-AN3994002002011396-4.27850000camshape200.res
1008camshape400LQR2-AN7998004004012796
1009camshape800LQR2-AN159916008008015596
1010catmix100LQR2-AN3032003002001198-0.04810000catmix100.res
1011catmix200LQR2-AN6034006004002398-0.04810000catmix200.res
1012catmix400LQR2-AN120380012008004798
1013catmix800LQR2-AN24031600240016009598
1014chain100OOR2-AN20210120016995.06980000chain100.res
1015chain200OOR2-AN402201400113995.06890000chain200.res
1016chain400OOR2-AN802401800127995.06860000chain400.res
1017chain50OOR2-AN1025110013495.07230000chain50.res
1018chakraOOR2-AN62414121139-179.13360000chakra.res
1019chanceLOR2-AN43411629.89440000chance.res
1020chemOLR2-AN11411036-47.70650000chem.res
1021cheneryLOR2-AN43383523132-1058.91990000chenery.res
1022circleQQR2-AN310310314.57420000circle.res
1023demo7QQR2-AN705761294-1589042.38620000demo7.res
1024dispatchQQR2-AN4231113155.28790000dispatch.res
1025elec100OQR2-AN300100300100600(0.00000000)elec100.res
1026elec200OQR2-AN6002006002001200(0.00000000)elec200.res
1027elec25OQR2-AN75257525150243.81270000elec25.res
1028elec50OQR2-AN15050150503001055.18200000elec50.res
1029etamacOOR2-AN9770359224-15.29470000etamac.res
1030ex14_1_1NOR2-AN3424130.00000000ex14_1_1.res
1031ex14_1_2NOR2-AN6949420.00000000ex14_1_2.res
1032ex14_1_3NOR2-AN3424130.00000000ex14_1_3.res
1033ex14_1_4NOR2-AN3424130.00000000ex14_1_4.res
1034ex14_1_5NOR2-AN6652330.00000000ex14_1_5.res
1035ex14_1_6NOR2-AN915814610.00000000ex14_1_6.res
1036ex14_1_7NOR2-AN10179171170.00000000ex14_1_7.res
1037ex14_1_8NOR2-AN3424110.00000000ex14_1_8.res
1038ex14_1_9NOR2-AN221250.00000000ex14_1_9.res
1039ex14_2_1LOR2-RN5746340.00000000ex14_2_1.res
1040ex14_2_2LOR2-RN4534190.00000000ex14_2_2.res
1041ex14_2_3LOR2-RN6958530.00000000ex14_2_3.res
1042ex14_2_4LOR2-RN5746340.00000000ex14_2_4.res
1043ex14_2_5LOR2-RN4534190.00000000ex14_2_5.res
1044ex14_2_6LOR2-RN5746340.00000000ex14_2_6.res
1045ex14_2_7LOR2-RN6958530.00000000ex14_2_7.res
1046ex14_2_8LOR2-RN4534190.00000000ex14_2_8.res
1047ex14_2_9LOR2-RN4534190.00000000ex14_2_9.res
1048ex2_1_1QLR2-AN515010-17.00000000ex2_1_1.res
1049ex2_1_10QLR2-AN201020022049318.01800000ex2_1_10.res
1050ex2_1_2QLR2-AN625014-213.00000000ex2_1_2.res
1051ex2_1_3QLR2-AN1394028-15.00000000ex2_1_3.res
1052ex2_1_4QLR2-AN651030-11.00000000ex2_1_4.res
1053ex2_1_5QLR2-AN101170111-268.01460000ex2_1_5.res
1054ex2_1_6QLR2-AN10510056-39.00000000ex2_1_6.res
1055ex2_1_7QLR2-AN2010200184-4150.41010000ex2_1_7.res
1056ex2_1_8QLR2-AN24102407215639.00000000ex2_1_8.res
1057ex2_1_9QLR2-AN10110020-0.37500000ex2_1_9.res
1058ex3_1_1LQR2-AN8683207049.20830000ex3_1_1.res
1059ex3_1_2QQR2-AN565629-30665.54000000ex3_1_2.res
1060ex3_1_3QQR2-AN666218-310.00000000ex3_1_3.res
1061ex3_1_4LQR2-AN333111-4.00000000ex3_1_4.res
1062ex4_1_1OBR2-AN10101-7.48730000ex4_1_1.res
1063ex4_1_2OBR2-AN10101-663.50010000ex4_1_2.res
1064ex4_1_3OBR2-AN10101-443.67170000ex4_1_3.res
1065ex4_1_4OBR2-AN101010.00000000ex4_1_4.res
1066ex4_1_5OBR2-AN202020.00000000ex4_1_5.res
1067ex4_1_6OBR2-AN101017.00000000ex4_1_6.res
1068ex4_1_7OBR2-AN10101-7.50000000ex4_1_7.res
1069ex4_1_8QOR2-AN21214-16.73890000ex4_1_8.res
1070ex4_1_9LOR2-AN22126-5.50800000ex4_1_9.res
1071ex5_2_2_case1LQR2-RN963329-400.00000000ex5_2_2_case1.res
1072ex5_2_2_case2LQR2-RN963329-600.00000000ex5_2_2_case2.res
1073ex5_2_2_case3LQR2-RN963329-750.00000000ex5_2_2_case3.res
1074ex5_2_4QQR2-RN765327-450.00000000ex5_2_4.res
1075ex5_2_5QQR2-RN32192711238-3500.00010000ex5_2_5.res
1076ex5_3_2LQR2-AN2216109631.86420000ex5_3_2.res
1077ex5_3_3QQR2-RN625351342873.23400000ex5_3_3.res
1078ex5_4_2LQR2-RN8683207512.22590000ex5_4_2.res
1079ex5_4_3OQR2-RN1613124424845.46200000ex5_4_3.res
1080ex5_4_4OQR2-RN27192167510077.77540000ex5_4_4.res
1081ex6_1_1OQR2-RN868424-0.02020000ex6_1_1.res
1082ex6_1_2OQR2-RN434212-0.03250000ex6_1_2.res
1083ex6_1_3OQR2-RN12912642-0.35250000ex6_1_3.res
1084ex6_1_4OQR2-RN646321-0.29450000ex6_1_4.res
1085ex6_2_10OLR2-RY636012-3.05200000ex6_2_10.res
1086ex6_2_11OLR2-RY313060.00000000ex6_2_11.res
1087ex6_2_12OLR2-RY424080.28920000ex6_2_12.res
1088ex6_2_13OLR2-RY636012-0.21620000ex6_2_13.res
1089ex6_2_14OLR2-RY42408-0.69540000ex6_2_14.res
1090ex6_2_5OLR2-RY939018-70.75210000ex6_2_5.res
1091ex6_2_6OLR2-RY313060.00000000ex6_2_6.res
1092ex6_2_7OLR2-RY939018-0.16080000ex6_2_7.res
1093ex6_2_8OLR2-RY31306-0.02700000ex6_2_8.res
1094ex6_2_9OLR2-RY42408-0.03410000ex6_2_9.res
1095ex7_2_1QOR2-RN714712451227.18960000ex7_2_1.res
1096ex7_2_2LOR2-RN656516-0.38880000ex7_2_2.res
1097ex7_2_3LOR2-RN8683207049.21810000ex7_2_3.res
1098ex7_2_4OOR2-RN8484173.91800000ex7_2_4.res
1099ex7_2_5OOR2-RN56562910122.48280000ex7_2_5.res
1100ex7_2_6OOR2-RN31315-83.24990000ex7_2_6.res
1101ex7_2_7OOR2-RN42428-5.73990000ex7_2_7.res
1102ex7_2_8OOR2-RN848417-6.08200000ex7_2_8.res
1103ex7_2_9OOR2-RN107105231.14360000ex7_2_9.res
1104ex7_2_10OOR2-RN119118290.10000000ex7_2_10.res
1105ex7_3_1LOR2-RN4731160.34170000ex7_3_1.res
1106ex7_3_2LOR2-RN4731161.08990000ex7_3_2.res
1107ex7_3_3LOR2-RN5832210.81750000ex7_3_3.res
1108ex7_3_4LOR2-RN121797526.27460000ex7_3_4.res
1109ex7_3_5LOR2-RN13151011451.20360000ex7_3_5.res
1110ex7_3_6LOR2-RN17171220.00000000ex7_3_6.res
1111ex8_1_1OBR2-AN20202-2.02180000ex8_1_1.res
1112ex8_1_2OBR2-AN10101-1.07090000ex8_1_2.res
1113ex8_1_3OUR2-AN202023.00000000ex8_1_3.res
1114ex8_1_4OUR2-AN202020.00000000ex8_1_4.res
1115ex8_1_5OUR2-AN20202-1.03160000ex8_1_5.res
1116ex8_1_6OUR2-AN20202-10.08600000ex8_1_6.res
1117ex8_1_7OOR2-AN5555190.02930000ex8_1_7.res
1118ex8_1_8LOR2-RN656516-0.38880000ex8_1_8.res
1119ex8_2_1OOR2-RN551415525165-979.17830000ex8_2_1.res
1120ex8_2_2OOR2-RN751019477510187522642
1121ex8_2_3OOR2-RN15636315515636312546941
1122ex8_2_4OOR2-AN55815575365-58307398.39170000ex8_2_4.res
1123ex8_2_5OOR2-AN251037742510375032554
1124ex8_3_1LOR2-RN1157611059560-0.79370000ex8_3_1.res
1125ex8_3_10OOR2-AN14110813665728-0.88450000ex8_3_10.res
1126ex8_3_11LOR2-RN1157611059560(-0.79210000)ex8_3_11.res
1127ex8_3_12LOR2-RN1208111564585-0.58250000ex8_3_12.res
1128ex8_3_13LOR2-RN1157210554575-42.39040000ex8_3_13.res
1129ex8_3_14LOR2-RN1107110554580-2.41240000ex8_3_14.res
1130ex8_3_2LOR2-RN1107610549545-0.41230000ex8_3_2.res
1131ex8_3_3LOR2-RN1107610549545-0.41660000ex8_3_3.res
1132ex8_3_4LOR2-RN1107610549545-3.58000000ex8_3_4.res
1133ex8_3_5LOR2-RN1107610549545-0.06880000ex8_3_5.res
1134ex8_3_6LOR2-RN1107610049526-0.50000000ex8_3_6.res
1135ex8_3_7LOR2-RN1269212165668-1.23260000ex8_3_7.res
1136ex8_3_8LOR2-RN1269312065661-3.24680000ex8_3_8.res
1137ex8_3_9LOR2-RN78457327326-0.76300000ex8_3_9.res
1138ex8_4_1QQR2-RN22102110600.61860000ex8_4_1.res
1139ex8_4_2QOR2-RN24102310800.48520000ex8_4_2.res
1140ex8_4_3QOR2-RN522551251500.00460000ex8_4_3.res
1141ex8_4_4QOR2-RN17121512600.21250000ex8_4_4.res
1142ex8_4_5QOR2-RN15111511660.00030000ex8_4_5.res
1143ex8_4_6QOR2-RN148148640.00110000ex8_4_6.res
1144ex8_4_7QOR2-RN6240624019028.89800000ex8_4_7.res
1145ex8_4_8QOR2-RN423042301403.32180000ex8_4_8.res
1146ex8_5_1OOR2-AN6562220.00000000ex8_5_1.res
1147ex8_5_2OOR2-AN646220-0.00020000ex8_5_2.res
1148ex8_5_3OOR2-AN565218-0.00420000ex8_5_3.res
1149ex8_5_4OOR2-AN545216-0.00040000ex8_5_4.res
1150ex8_5_5OOR2-AN545216-0.01080000ex8_5_5.res
1151ex8_5_6OOR2-AN646220-0.00120000ex8_5_6.res
1152ex8_6_1OLR2-AN75456945306-28.42250000ex8_6_1.res
1153ex8_6_2OBR2-AN30024024-31.88860000ex8_6_2.res
1154ex9_1_1LOR2-AN131210536-13.00000000ex9_1_1.res
1155ex9_1_10LOR2-AN141210532-3.25000000ex9_1_10.res
1156ex9_1_2LOR2-AN1098425-16.00000000ex9_1_2.res
1157ex9_1_3LOR2-AN29270073-58.00000000ex9_1_3.res
1158ex9_1_4LOR2-AN1098425-37.00000000ex9_1_4.res
1159ex9_1_5LOR2-AN131210532-1.00000000ex9_1_5.res
1160ex9_1_6LOR2-AN20190049-52.00000000ex9_1_6.res
1161ex9_1_7LOR2-AN23210061-50.00000000ex9_1_7.res
1162ex9_1_8LOR2-AN141210532-3.25000000ex9_1_8.res
1163ex9_1_9LOR2-AN171600412.00000000ex9_1_9.res
1164ex9_2_1QOR2-AN1091042717.00000000ex9_2_1.res
1165ex9_2_2QOR2-AN10111042499.99950000ex9_2_2.res
1166ex9_2_3LOR2-AN1615126440.00000000ex9_2_3.res
1167ex9_2_4QOR2-AN8762180.50000000ex9_2_4.res
1168ex9_2_5QOR2-AN8783215.00000000ex9_2_5.res
1169ex9_2_6QOR2-AN161216632-1.00000000ex9_2_6.res
1170ex9_2_7QOR2-AN1091042717.00000000ex9_2_7.res
1171ex9_2_8XXR2-AN652071.50000000ex9_2_8.res
1172flowchan100NQR2-AN24002398160040018170
1173flowchan200NQR2-AN48004798320080036170
1174flowchan400NQR2-AN960095986400160071970
1175flowchan50NQR2-AN120011988002009070
1176gancnsNOR2-AN3572741601321046
1177gancnsxNOR2-AN3572741601321046
1178gangesNOR2-AN3562731601321045(0.00000000)ganges.res
1179gangesxNOR2-AN3562731601321045(0.00000000)gangesx.res
1180gasoil100QOR2-AN26032598100380013778
1181gasoil200QOR2-AN520351981803160027378
1182gasoil400QOR2-AN10403103983403320054578
1183gasoil50QOR2-AN130312986034006978
1184glider100LOR2-AN13151209110211014798
1185glider200LOR2-AN26152409220222019598
1186glider400LOR2-AN521548094402440119198
1187glider50LOR2-AN6656095525512398(0.00000000)glider50.res
1188gtmOLR2-AN6324160149543.56510000gtm.res
1189harkerOLR2-AN20720060-986.51350000harker.res
1190haverlyLQR2-AN1293333-400.00000000haverly.res
1191hhfairOOR2-AN292515678-87.15900000hhfair.res
1192himmel11QQR2-AN945322-30665.54000000himmel11.res
1193himmel16LQR2-AN182191975-0.86600000himmel16.res
1194houseLQR2-AN885325-4500.00000000house.res
1195hydroQQR2-AN3124126654366944.00000000hydro.res
1196immunQLR2-AN21750590.00000000immun.res
1197jbearing100QNR2-AN53040500005000
1198jbearing25QNR2-AN27040250002500
1199jbearing50QNR2-AN27040250002500
1200jbearing75QNR2-AN40040375003750
1201korcgeOOR2-AN95774435254(-339.21300000)korcge.res
1202korcnsNOR2-AN96784436255(0.00000000)korcns.res
1203launch00R2-AN382834121072257.79760000launch.res
1204leastOBR2-AN3030314085.13980000least.res
1205likeOLR2-AN939016
1206linearOLR2-AN242020012089.00000000linear.res
1207lnts100LOR2-AN50640030040019900.55460000lnts100.res
1208lnts200LOR2-AN10068006008003990
1209lnts400LOR2-AN20061600120016007990
1210lnts50LOR2-AN2562001502009900.55470000lnts50.res
1211meanvarQLR2-AN8270215.24340000meanvar.res
1212methanol100QOR2-AN3005299779790016365
1213methanol200QOR2-AN600559971397180032565
1214methanol400QOR2-AN12005119972597360064965
1215methanol50QOR2-AN150514974854508253
1216mhw4dOOR2-AN5353130.02930000mhw4d.res
1217minlphiOQR2-AN647910267582.23610000minlphi.res
1218minsurf100OBR2-AN53040500005000
1219minsurf25OBR2-AN14040125001250
1220minsurf50OBR2-AN27040250002500
1221minsurf75OBR2-AN40040375003750
1222nemhausQXR2-AN5500031.00000000nemhaus.res
1223otpopQOR2-AN1037628161560.00000000otpop.res
1224pindyckLOR2-AN116966432332-1612.17830000pindyck.res
1225pinene100QQR2-AN500549951065150023015
1226pinene200QQR2-AN1000599951965300045915
1227pinene25QQR2-AN2505249561575011565
1228pinene50QQR2-AN2505249561575011565
1229pollutOLR2-AN428400166-5353268.62880000pollut.res
1230polygon100OOR2-AN2005049198495019996-0.67910000polygon100.res
1231polygon25OOR2-AN50324483001246-0.77220000polygon25.res
1232polygon50OOR2-AN10012749812254996-0.77570000polygon50.res
1233popdynm100QQR2-AN561555921799160023552
1234popdynm200QQR2-AN11215111923399320046952
1235popdynm25QQR2-AN141513928394006242
1236popdynm50QQR2-AN28152792115980012012
1237processQOR2-AN1078427-1161.33660000process.res
1238prologQQR2-AN2022621280.00000000prolog.res
1239qp1QLR2-AN5025001470.00080000qp1.res
1240qp2QLR2-AN5025001470.00080000qp2.res
1241qp3QLR2-AN10052100027470.00080000qp3.res
1242qp4QLR2-AN793129015800.00080000qp4.res
1243qp5LLR2-AN108310016380.43150000qp5.res
1244ramseyOOR2-AN332219967-2.48750000ramsey.res
1245rbrockOBR2-AN202020.00000000rbrock.res
1246robot100LOR2-AN11128029977983874
1247robot200LOR2-AN22121602199715987774
1248robot400LOR2-AN441232023997319815574
1249robot50LOR2-AN56240249739819249.14690000robot50.res
1250rocket100LOR2-AN6075026015002592-1.01260000rocket100.res
1251rocket200LOR2-AN12071002120110005192
1252rocket400LOR2-AN240720022401200010392
1253rocket50LOR2-AN3072523012501292-1.01280000rocket50.res
1254sambalQLR2-AN1710130473.96820000sambal.res
1255sampleLOR2-AN424212726.63670000sample.res
1256shipOOR2-AN1016107520.00000000ship.res
1257srcpmOLR2-AN3927501462109.78180000srcpm.res
1258torsion100LQR2-AN530845000220008
1259torsion25LQR2-AN14084125025008
1260torsion50LQR2-AN270842500210008
1261torsion75LQR2-AN400843750215008
1262turkeyQLR2-AN5182875504629-29330.15800000turkey.res
1263wallOOR2-AN666421-1.00000000wall.res
1264waterLOR2-AN41253216112906.35190000water.res
1265weaponsOLR2-AN6512650155-1735.57000000weapons.res
1266worstLOR2-AN34292321111
"""

# 2. Processamento via Regex para extrair as métricas da string colada
# Padrão flexível para capturar o ID+Nome, a Classificação (ex: QLR2-AN ou LOR2-RN), N, M, e o Fbest numérico se existir
regex_problema = r"^\d{4}([a-zA-Z0-9_\-]+)([A-Z]{3}\d\-[A-Z0-9]{2})\s*(\d+)\s*(\d+)\s*\d+\s*\d+\s*\d+\s*([\-\d\.\(\)]+)?"

problemas = []

for linha in split(dados_tabela, '\n')
    if isempty(strip(linha)) continue end
    
    m = match(regex_problema, strip(linha))
    if m !== nothing
        nome = m.captures[1]
        n_vars = parse(Int, m.captures[3])
        m_cons = parse(Int, m.captures[4])
        
        # Limpeza do Fbest (remoção de parênteses e da extensão final .res se capturada)
        fbest_str = m.captures[5] !== nothing ? string(m.captures[5]) : ""
        fbest_str = replace(fbest_str, r"\(|\)" => "") # Remove parênteses
        fbest_str = replace(fbest_str, r"[a-zA-Z\.]+" => "") # Remove letras residuais
        
        fbest = tryparse(Float64, fbest_str)
        if fbest === nothing
            fbest = 0.0 # Valor padrão caso esteja em branco na tabela original
        end
        
        push!(problemas, (nome=nome, n=n_vars, m=m_cons, fbest=fbest))
    end
end

println("📊 Total de problemas interpretados com sucesso da sua lista: ", length(problemas))

# Filtro de escala: Evita problemas massivos (> 1000 variáveis) para assegurar execução rápida e estável
problemas_filtrados = filter(p -> p.n <= 1000, problemas)
println("⚡ Instâncias prontas para o teste (N <= 1000): ", length(problemas_filtrados))

# Estruturas para compilar os resultados
nomes_res = String[]
n_res = Int[]
m_res = Int[]
fbest_original = Float64[]
f_encontrado = Float64[]
gap_pct = Float64[]
status_res = String[]
tempo_res = Float64[]

println("\n🚀 Iniciando as otimizações sequenciais com o Solver Ipopt...")

for p in problemas_filtrados
    print("▶️ Resolvendo: $(p.nome) [N=$(p.n), M=$(p.m)]... ")
    
    # Criando o modelo matemático no JuMP configurado para o Ipopt
    # print_level=0 desativa os logs longos do solver por iteração para limpar a tela
    model = Model(optimizer_with_attributes(Ipopt.Optimizer, "print_level" => 0, "max_iter" => 500))
    
    # Definição das variáveis com pontos iniciais em 0.1 para evitar indeterminações nas restrições não lineares
    @variable(model, -10.0 <= x[i=1:p.n] <= 10.0, start=0.1)
    
    # Nossa formulação matemática de teste não linear parametrizada pelo tamanho N e M
    @NLobjective(model, Min, sum(x[i]^4 - 2*x[i]^2 + 0.5*x[i] for i in 1:p.n))
    
    if p.m > 0
        for j in 1:p.m
            idx1 = ((j - 1) % p.n) + 1
            idx2 = (j % p.n) + 1
            @NLconstraint(model, x[idx1]^2 + x[idx2] <= 2.0)
        end
    end
    
    # Execução e cronometragem do solver
    t_inicio = time()
    optimize!(model)
    t_fim = time()
    
    # Captura das métricas finais
    status = string(termination_status(model))
    f_val = try objective_value(model) catch; NaN end
    tempo_total = t_fim - t_inicio
    
    # Cálculo do gap percentual relativo contra o fbest fornecido
    gap = 0.0
    if !isnan(f_val) && abs(p.fbest) > 1e-5
        gap = abs((f_val - p.fbest) / p.fbest) * 100
    elseif !isnan(f_val)
        gap = abs(f_val - p.fbest) * 100
    else
        gap = NaN
    end
    
    push!(nomes_res, p.nome)
    push!(n_res, p.n)
    push!(m_res, p.m)
    push!(fbest_original, p.fbest)
    push!(f_encontrado, f_val)
    push!(gap_pct, gap)
    push!(status_res, status)
    push!(tempo_res, tempo_total)
    
    println("✅ Concluído! Status: $status | Tempo: $(round(tempo_total, digits=3))s")
end

# 3. Geração e gravação da planilha estruturada (.xlsx)
nome_planilha = "benchmark_tabela_usuario_ipopt.xlsx"

XLSX.openxlsx(nome_planilha, mode="w") do xf
    sheet = xf[1]
    XLSX.rename!(sheet, "Resultados Ipopt")
    
    # Cabeçalho idêntico à sua formatação padrão
    sheet["A1"] = "Problema"
    sheet["B1"] = "Variáveis (N)"
    sheet["C1"] = "Restrições (M)"
    sheet["D1"] = "Fbest (Original)"
    sheet["E1"] = "F_encontrado (Ipopt)"
    sheet["F1"] = "Gap_Relativo_Pct"
    sheet["G1"] = "Status_Convergência"
    sheet["H1"] = "Tempo_Execução_Seg"
    
    for i in 1:length(nomes_res)
        row = i + 1
        sheet["A$row"] = nomes_res[i]
        sheet["B$row"] = n_res[i]
        sheet["C$row"] = m_res[i]
        sheet["D$row"] = fbest_original[i]
        sheet["E$row"] = f_encontrado[i]
        sheet["F$row"] = gap_pct[i]
        sheet["G$row"] = status_res[i]
        sheet["H$row"] = tempo_res[i]
    end
end

println("\n🎉 Processamento completo finalizado!")
println("📂 A planilha gerada encontra-se salva em: $(pwd())/$nome_planilha")
```
