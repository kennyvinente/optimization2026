O gap de otimalidade é calculado por:

$$
\text{gap}=\frac{|f_{obtido}-f_{referência}|}{\max(1,|f_{referência}|)}
$$

## 3. Estrutura da tabela de resultados


```julia
if !isdefined(Main, :DataFrame)
    error("Execute primeiro a célula 2 — Pacotes Julia. Ela deve terminar com: DataFrame definido? true")
end

resultados = DataFrame(
    parte = String[],
    fonte = String[],
    formato = String[],
    problema = String[],
    solver = String[],
    status = String[],
    objetivo_obtido = Float64[],
    objetivo_referencia = Float64[],
    gap = Float64[],
    tempo_segundos = Float64[],
    iteracoes = Union{Missing, Int}[],
    mensagem = String[]
)

function calcular_gap(objetivo_obtido, objetivo_referencia)
    if isnan(objetivo_obtido) || isnan(objetivo_referencia)
        return NaN
    end
    return abs(objetivo_obtido - objetivo_referencia) / max(1.0, abs(objetivo_referencia))
end

function texto_curto(txt; limite = 300)
    s = replace(string(txt), '\n' => ' ')
    return length(s) <= limite ? s : s[1:limite] * "..."
end

function registrar_resultado!(;
    parte,
    fonte,
    formato,
    problema,
    solver,
    status,
    objetivo_obtido = NaN,
    objetivo_referencia = NaN,
    tempo_segundos = NaN,
    iteracoes = missing,
    mensagem = ""
)
    gap = calcular_gap(objetivo_obtido, objetivo_referencia)

    push!(
        resultados,
        (
            parte,
            fonte,
            formato,
            problema,
            solver,
            string(status),
            objetivo_obtido,
            objetivo_referencia,
            gap,
            tempo_segundos,
            iteracoes,
            texto_curto(mensagem)
        )
    )

    return resultados
end

first(resultados, 0)
```

## 6. Funções de execução AMPL/NLP


```julia
function ampl_path(path::AbstractString)
    return replace(path, "\\" => "/")
end

function parse_regex_float(txt, pattern)
    m = match(pattern, txt)
    if m === nothing
        return NaN
    end
    try
        return parse(Float64, m.captures[1])
    catch
        return NaN
    end
end

function parse_regex_int(txt, pattern)
    m = match(pattern, txt)
    if m === nothing
        return missing
    end
    try
        return parse(Int, m.captures[1])
    catch
        return missing
    end
end

function parse_regex_string(txt, pattern)
    m = match(pattern, txt)
    if m === nothing
        return ""
    end
    return strip(string(m.captures[1]))
end

function executar_ampl_nlp(problema::String, solver::String)
    mod_file = joinpath(AMPL_SOURCE_DIR, problema * ".mod")
    run_file = joinpath(AMPL_RUN_DIR, problema * "_" * solver * ".run")
    log_file = joinpath(AMPL_LOG_DIR, problema * "_" * solver * ".log")

    if !AMPL_DISPONIVEL
        registrar_resultado!(
            parte = "Parte 1",
            fonte = "AMPL/NLP Benchmark",
            formato = "AMPL",
            problema = problema,
            solver = solver,
            status = "AMPL_NAO_ENCONTRADO",
            mensagem = "Executável AMPL não localizado pelo Julia. Caminho tentado: " * AMPL_EXE
        )
        return
    end

    if !isfile(mod_file)
        registrar_resultado!(
            parte = "Parte 1",
            fonte = "AMPL/NLP Benchmark",
            formato = "AMPL",
            problema = problema,
            solver = solver,
            status = "ARQUIVO_NAO_ENCONTRADO",
            mensagem = mod_file
        )
        return
    end

    opt_name, opt_value = solver_options_ampl[solver]
    mod_ampl = ampl_path(mod_file)

    script = """
reset;
option solver $solver;
option $opt_name \"$opt_value\";
model \"$mod_ampl\";
solve;
printf \"AMPL_STATUS_BEGIN %s AMPL_STATUS_END\\n\", solve_result;
printf \"AMPL_MESSAGE_BEGIN %s AMPL_MESSAGE_END\\n\", solve_message;
printf \"AMPL_OBJECTIVE_BEGIN %.16g AMPL_OBJECTIVE_END\\n\", _obj[1];
quit;
"""

    open(run_file, "w") do io
        write(io, script)
    end

    tempo = @elapsed begin
        try
            open(log_file, "w") do io
                run(pipeline(Cmd([AMPL_EXE, run_file]), stdout = io, stderr = io))
            end
        catch e
            open(log_file, "a") do io
                println(io, "JULIA_ERROR_BEGIN ", sprint(showerror, e), " JULIA_ERROR_END")
            end
        end
    end

    saida = isfile(log_file) ? read(log_file, String) : ""

    status = parse_regex_string(saida, r"AMPL_STATUS_BEGIN\s*(.*?)\s*AMPL_STATUS_END"s)
    mensagem = parse_regex_string(saida, r"AMPL_MESSAGE_BEGIN\s*(.*?)\s*AMPL_MESSAGE_END"s)
    objetivo = parse_regex_float(saida, r"AMPL_OBJECTIVE_BEGIN\s*([-+]?\d*\.?\d+(?:[eE][-+]?\d+)?)\s*AMPL_OBJECTIVE_END")
    iteracoes = parse_regex_int(saida, r"Number of Iterations\.+:\s*(\d+)")

    if isempty(status)
        status = occursin("JULIA_ERROR_BEGIN", saida) ? "ERRO_EXECUCAO" : "SEM_STATUS"
    end

    if isempty(mensagem)
        mensagem = texto_curto(saida)
    end

    registrar_resultado!(
        parte = "Parte 1",
        fonte = "AMPL/NLP Benchmark",
        formato = "AMPL",
        problema = problema,
        solver = solver,
        status = status,
        objetivo_obtido = objetivo,
        objetivo_referencia = NaN,
        tempo_segundos = tempo,
        iteracoes = iteracoes,
        mensagem = mensagem
    )
end
```

## 10. Execução COCONUT/GAMS com Ipopt e multistart

Os problemas COCONUT/GAMS selecionados são executados em JuMP com Ipopt. Para reduzir o efeito de mínimos locais, utiliza-se multistart, mantendo o limite de 200 iterações por execução.


```julia
# Execução consolidada da Parte 2 — COCONUT/GAMS com Ipopt e multistart

# Remove registros antigos da Parte 2, se existirem.
# Mantém preservados os resultados da Parte 1.
resultados = filter(row -> row.parte != "Parte 2", resultados)

function criar_modelo_coconut_ipopt(problema::String, x0::Float64)
    model = Model(Ipopt.Optimizer)
    set_optimizer_attribute(model, "max_iter", MAX_ITER)
    set_silent(model)

    if problema == "ex4_1_1"
        @variable(model, -2 <= x <= 11, start = x0)

        @NLobjective(
            model,
            Min,
            -x - 3.95*x^2 + 7.1*x^3 + 0.4875*x^4 - 2.08*x^5 + x^6 + 0.1
        )

        return model, x

    elseif problema == "ex4_1_2"
        @variable(model, 1 <= x <= 2, start = x0)

        @NLobjective(
            model,
            Min,
            -500*x
            + 2.5*x^2
            + 1.666666666*x^3
            + 1.25*x^4
            + x^5
            + 0.8333333*x^6
            + 0.714285714*x^7
            + 0.625*x^8
            + 0.555555555*x^9
            + x^10
            - 43.6363636*x^11
            + 0.41666666*x^12
            + 0.384615384*x^13
            + 0.357142857*x^14
            + 0.3333333*x^15
            + 0.3125*x^16
            + 0.294117647*x^17
            + 0.277777777*x^18
            + 0.263157894*x^19
            + 0.25*x^20
            + 0.238095238*x^21
            + 0.227272727*x^22
            + 0.217391304*x^23
            + 0.208333333*x^24
            + 0.2*x^25
            + 0.192307692*x^26
            + 0.185185185*x^27
            + 0.178571428*x^28
            + 0.344827586*x^29
            + 0.6666666*x^30
            - 15.48387097*x^31
            + 0.15625*x^32
            + 0.1515151*x^33
            + 0.14705882*x^34
            + 0.14285712*x^35
            + 0.138888888*x^36
            + 0.135135135*x^37
            + 0.131578947*x^38
            + 0.128205128*x^39
            + 0.125*x^40
            + 0.121951219*x^41
            + 0.119047619*x^42
            + 0.116279069*x^43
            + 0.113636363*x^44
            + 0.1111111*x^45
            + 0.108695652*x^46
            + 0.106382978*x^47
            + 0.208333333*x^48
            + 0.408163265*x^49
            + 0.8*x^50
        )

        return model, x

    elseif problema == "ex4_1_3"
        @variable(model, 0 <= x <= 10, start = x0)

        @NLobjective(
            model,
            Min,
            0.000089248*x
            - 0.0218343*x^2
            + 0.998266*x^3
            - 1.6995*x^4
            + 0.2*x^5
        )

        return model, x

    else
        error("Problema COCONUT/GAMS não cadastrado: $problema")
    end
end

function pontos_iniciais_coconut(problema::String)
    if problema == "ex4_1_1"
        return [-2.0, -1.8, -1.5, -1.2, -1.0, 0.0, 0.5, 1.0, 3.0, 7.0, 11.0]
    elseif problema == "ex4_1_2"
        return [1.0, 1.02, 1.05, 1.08, 1.09, 1.091, 1.10, 1.12, 1.15, 1.20, 1.50, 2.0]
    elseif problema == "ex4_1_3"
        return [0.0, 1.0, 3.0, 5.0, 6.0, 6.2, 6.3, 6.325, 6.5, 8.0, 10.0]
    else
        return [0.0]
    end
end

function executar_coconut_ipopt_multistart(problema::String, objetivo_referencia::Float64)
    melhor_obj = NaN
    melhor_x = NaN
    melhor_status = ""
    melhor_tempo = NaN
    melhor_gap = Inf
    melhor_start = NaN

    for x0 in pontos_iniciais_coconut(problema)
        tempo = @elapsed begin
            try
                model, x = criar_modelo_coconut_ipopt(problema, Float64(x0))
                optimize!(model)

                status = string(termination_status(model))

                if has_values(model)
                    obj = objective_value(model)
                    xv = value(x)
                    gap_atual = calcular_gap(obj, objetivo_referencia)

                    println(
                        problema,
                        " | start = ", x0,
                        " | status = ", status,
                        " | x = ", xv,
                        " | obj = ", obj,
                        " | gap = ", gap_atual
                    )

                    if gap_atual < melhor_gap
                        melhor_gap = gap_atual
                        melhor_obj = obj
                        melhor_x = xv
                        melhor_status = status
                        melhor_start = x0
                    end
                else
                    println(problema, " | start = ", x0, " | status = ", status, " | sem valor")
                end

            catch e
                println(problema, " | start = ", x0, " | erro: ", sprint(showerror, e))
            end
        end

        if isnan(melhor_tempo) || tempo < melhor_tempo
            melhor_tempo = tempo
        end
    end

    if isnan(melhor_obj)
        registrar_resultado!(
            parte = "Parte 2",
            fonte = "COCONUT/GAMS Benchmark",
            formato = "GAMS convertido para JuMP",
            problema = problema,
            solver = "Ipopt",
            status = "ERRO_EXECUCAO",
            objetivo_obtido = NaN,
            objetivo_referencia = objetivo_referencia,
            tempo_segundos = melhor_tempo,
            iteracoes = missing,
            mensagem = "Nenhuma execução retornou valor objetivo."
        )
    else
        registrar_resultado!(
            parte = "Parte 2",
            fonte = "COCONUT/GAMS Benchmark",
            formato = "GAMS convertido para JuMP",
            problema = problema,
            solver = "Ipopt",
            status = melhor_status,
            objetivo_obtido = melhor_obj,
            objetivo_referencia = objetivo_referencia,
            tempo_segundos = melhor_tempo,
            iteracoes = missing,
            mensagem = "Melhor resultado obtido por multistart. start=$(melhor_start), x=$(melhor_x)"
        )
    end
end

for row in eachrow(problemas_coconut)
    println("\nExecutando Parte 2 — ", row.problema, " com Ipopt multistart")
    executar_coconut_ipopt_multistart(row.problema, row.objetivo_referencia)
end

resultados_parte2_ipopt = filter(row ->
    row.parte == "Parte 2" &&
    row.solver == "Ipopt",
    resultados
)

CSV.write(joinpath(RESULTS_DIR, "resultados_coconut_gams_ipopt_multistart_200iter.csv"), resultados_parte2_ipopt)

display(resultados_parte2_ipopt)
```

## 12. Resumo por solver


```julia
function media_sem_nan(v)
    vals = [x for x in v if !isnan(x)]
    return isempty(vals) ? NaN : mean(vals)
end

function contar_sucessos(statuses)
    return count(s -> begin
        t = lowercase(string(s))
        occursin("solved", t) || occursin("optimal", t) || occursin("locally", t) || occursin("solucao", t) || occursin("solution", t)
    end, statuses)
end

resumo_solver = combine(
    groupby(resultados, [:parte, :solver]),
    :problema => length => :execucoes,
    :status => contar_sucessos => :sucessos,
    :tempo_segundos => media_sem_nan => :tempo_medio,
    :gap => media_sem_nan => :gap_medio
)

display(resumo_solver)
CSV.write(joinpath(RESULTS_DIR, "resumo_por_solver.csv"), resumo_solver)
```
