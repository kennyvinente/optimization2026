PROBLEMA DE PACKING

Importar as Bibliotecas

```julia
using JuMP, GLPK
```

Modelo

```julia
model = Model(GLPK.Optimizer)
```

Definir os Dados

```julia
peso = [4, 6, 5, 3]
valor = [10, 12, 8, 6]

n = length(peso)
capacidade = 10
```

Variáveis de Decisão

```julia
@variable(model, x[1:n], Bin)
```

Função Objetivo

```julia
@objective(model, Max, sum(valor[i] * x[i] for i in 1:n))
```

Restrição de Capacidade

```julia
@constraint(model,
    sum(peso[i] * x[i] for i in 1:n) <= capacidade
)
```

Resolver

```julia
optimize!(model)
```

Resultados

```julia
#Valor Ótimo

println("Valor máximo: ", objective_value(model))

#Itens selecionados
for i in 1:n
    if value(x[i]) > 0.5
        println("Item ", i, " selecionado")
    end
end
```
