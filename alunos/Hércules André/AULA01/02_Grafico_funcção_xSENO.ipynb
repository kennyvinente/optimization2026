Plot the graph of the function x.sin(x) over the interval [-10π, 10π] 

```julia
using Pkg
Pkg.add("Plots")
```

```julia
using Plots
```

Definir o Intervalo

```julia
x = range(-10π, 10π, length=1000)
```

Definir a função

```julia
y = x.* sin.(x)
```

Plotar o Gráfico

```julia
plot(x, y, label="x*sin(x)", xlabel="x", ylabel="f(x)", title="Gráfico de x*sin(x)")
```
