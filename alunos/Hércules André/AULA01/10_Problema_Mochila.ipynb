PROBLEMA DA MOCHILA

Importar as Bibliotecas

```julia
using JuMP, GLPK
```

Modelo

```julia
model = Model(GLPK.Optimizer)
```

Definir as variáveis binárias

```julia
@variable(model, x1 , Bin)
@variable(model, x2 , Bin)
@variable(model, x3 , Bin)
```

Função Objetivo

```julia
@objective(model, Max, 120x1 + 80x2 + 60x3)
```

Restrição de Peso

```julia
@constraint(model, 2x1 + x2 + x3 <= 3.5)
```

Resolver o problema

```julia
optimize!(model)
```

Resultados

```julia
println("Status:", termination_status(model))
println("Valor Máximo:", objective_value(model))

println("Tent:", value(x1))
println("Stove:", value(x2))
println("Food:", value(x3))
```
