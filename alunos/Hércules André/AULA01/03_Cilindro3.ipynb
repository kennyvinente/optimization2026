PROBLEMA DO CILINDRO

Importar as Bibliotecas

```julia
using JuMP, Ipopt
```

Modelo 

```julia
model=Model(Ipopt.Optimizer)
```

Definição das Variáveis

```julia
@variable(model, r >= 0)
@variable(model, h >= 0)
```

Definir os parâmetros

```julia
N = 10
c1 = 2
c2 = 0.5
```

Função Objetivo

```julia
@NLobjective(model,Min,2*c1*π*r^2 + 2*π*r*h*c2)
```

Restrição (Volume Fixo)

```julia
@NLconstraint(model, π*r^2*h == N)
```

Resolver

```julia
optimize!(model)
```

Escrever os Resultados

```julia
println("Raio ótimo (r): ", value(r))
println("Altura ótima (h): ", value(h))

println("Custo mínimo: ", objective_value(model))
```
