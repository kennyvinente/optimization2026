PROBLEMA DA DIETA

```julia
using JuMP, GLPK
```

CRIAR O MODELO

```julia
model = Model(GLPK.Optimizer)
```

DEFINIÇÃO DAS VARIÁVEIS

```julia
@variable(model, x1 >= 0) # Apple
@variable(model, x2 >= 0) # Bread
@variable(model, x3 >= 0) # Milk
@variable(model, x4 >= 0) # Egg
```

FUNÇÃO OBJETIVO (Minimizar custo)

```julia
@objective(model, Min, 1*x1 + 0.5*x2 + 2*x3 + 0.3*x4)
```

RESTRIÇÕES

```julia
#Calorias
@constraint(model, 100*x1 + 200*x2 + 150*x3 + 70*x4 >= 500)

#Proteína
@constraint(model, 0.5*x1 + 4*x2 + 8*x3 + 6*x4 >= 20)

#Vitaminas
@constraint(model, 2*x1 + 10*x3 >= 10)
```

RESOLVER O PROBLEMA

```julia
optimize!(model)
```

VER OS RESULTADOS

```julia
println("Status:", termination_status(model))
println("Custo mínimo = ", objective_value(model))

println("Apple = ", value(x1))
println("Bread = ", value(x2))
println("Milk = ", value(x3))
println("Egg = ", value(x4))
```
