Plot the graph of the function sin(x) over the interval   [-pi/4,3*pi/4]

```julia
using Pkg
Pkg.add("Plots")
```

Importar a Biblioteca

```julia
using Plots
```

Criar os intervalos

```julia
x = range(-π/4, 3π/4, length=100)
```

```julia
y = sin.(x)
```

Plotar os Graficos

```julia
plot( x, y, label="sin(x)", xlabel="x", ylabel="sin(x)", title = "Gráfico da Função Seno")
```
