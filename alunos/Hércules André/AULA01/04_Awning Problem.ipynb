Solve the Awning Problem considering the following data:
h : 2
w : 3
initial guess (x,y) = (1.0,1.0)

Maximizar:
        A=x⋅y

Sujeito a:
        x^2+(y−h)^2=w^2

INSTALAÇÃO DE PACOTES

```julia
using JuMP, Ipopt
```

DEFINIR OS DADOS

```julia
h = 2.0
w = 3.0
```

CRIAR O MODELO

```julia
model = Model(Ipopt.Optimizer)
```

DEFINIR AS VARIÁVEIS

```julia
@variable(model, x>=0)
@variable(model, y>=0)
```

FUNÇÃO OBJETIVO

```julia
@NLobjective(model, Max, x * y )
```

RESTRIÇÃO NÃO-LINEAR

```julia
@NLconstraint(model, x^2 + (y - h)^2 == w^2)
```

```julia
optimize!(model)
```

RESULTADO

```julia
x_opt = value(x)
y_opt = value(y)
area = x_opt * y_opt

println("x ótimo = ", x_opt)
println("y ótimo = ", y_opt)
println("Área máxima = ", area)
```
