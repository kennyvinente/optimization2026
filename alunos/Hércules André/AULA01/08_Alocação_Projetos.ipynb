PROBLEMA DE ALOCAÇÃO DE RECURSOS - Alocação de Designers em projetos

- Distribuir horas dos designers (1,2,3) entre os projetos (1,2,3,4)
- Maximar a pontuação total
- Respeitar:
      Cada designer tem 80 horas disponíveis
      Cada projeto de uma quantidade fixa de horas

IMPORTAR AS BIBLIOTECAS

```julia
using JuMP,GLPK 
```

CRIAR O MODELO

```julia
model=Model(GLPK.Optimizer)
```

DEFINIÇÃO DE DADOS

```julia
#Pontuações

c = [
    90 80 10 50;
    60 70 50 65;
    70 40 80 85]

# Horas requeridas por projeto
required = [70,50,85,35]

#conjuntos
designers = 1:3
projects = 1:4
```

```julia
@variable(model, x[designers, projects] >= 0)
```

FUNÇÃO OBJETIVO

```julia
@objective(model, Max, sum(c[i,j] for i in designers, j in projects))
```

RESTRIÇÕES

```julia
#Tempo dos designers

for i in designers
    @constraint(model, sum(x[i,j] for j in projects) <= 80)
end

#Demanda dos projetos

for j in projects
    @constraint(model, sum(x[i,j] for i in designers) == required[j])
end
```

RESOLVER

```julia
optimize!(model)
```

RESULTADO

```julia
#Valor Otimo
println("Valor ótimo:", objective_value(model))

#Alocação de horas
for i in designers, j in projects
    println("x[$i,$j] = ", value(x[i,j]))
end
```
